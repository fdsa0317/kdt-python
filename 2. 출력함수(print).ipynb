# 제어 문자
- 반드시 따옴표 안에서 작성한다.
  문자열 안에서 특수한 기능을 가지는 문자들을 의미한다.
- \n : new line, 개행 문자, 줄바꿈
- \t : tab, 위 아래 줄 간격 맞춰 띄기
- \" : "를 표현
- \' : '를 표현
- \\ : \를 표현

# 문자열
- 컴퓨터가 명령어가 아닌 문자열 값임을 구분할 수 있게 사용하는 특수한 기호
- 즉, 문자열은 문자들의 나열이다.

```python
# 함수: 이름뒤에 소괄호, 기능 또는 값
print("저의 이름은 이상아이고,\n나이는 26살입니다.")
```

```python
print("저의 이름은 \'이상아\'이고,\n나이는 \'26\'살입니다.")
```

## print(값, end='')
- print 함수가 종료될 때 처리하는 값을 제어하는 속성

```python
print("홍길동", end='')
print("입니다")
print("홍길동", end='\n\n')
print("입니다")
```

## doc-string(""" """)
- 모듈, 클래스, 함수, 첫 번째 줄을 설명하기 위한 문법
- 문자열 리터럴

```python
print("""
안녕하세요
홍길동입니다
""")
```

## 규칙
- 1. 내용 안에 들어가는 내용이라면 안으로 들여쓰기 해준다. [tab]
  2. 내용들끼리는 같은 depth(깊이)에 들여쓰기를 사용해야 한다.
     민약 들여쓰기를 잘못 넣었다면 [Shift] + [tab]으로 바깥으로 꺼내준다.
  3. 띄어쓰기(" ") -> 공백문자를 잘못 입력하면 depth로 인식한다.

```python
print("""
    안녕하세요
    홍길동입니다
""")
```
