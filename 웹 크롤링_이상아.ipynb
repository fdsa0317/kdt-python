```python
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/152.0.0.0 Safari/537.36"}
```

```python
import requests
from bs4 import BeautifulSoup as bs
import json

aladin_url = "https://www.aladin.co.kr/shop/common/wbest.aspx?BranchType=1"
response = requests.get(aladin_url, headers=headers)
aladin_soup = bs(response.text, "html.parser")
# aladin_soup

# 책 제목
titles = aladin_soup.select(".bo3")
title_list = list(map(lambda tag: tag.text, titles))
print(f"제목\n{title_list}")
print()

# 원가
price = aladin_soup.select(".ss_book_box table table .ss_book_list:first-child li")
price_list = [li.text.split("→")[0].strip().replace("원", "") for li in price[3::5]]
print(f"원가\n{price_list}")
print()

# 할인가
discount = aladin_soup.select("em")
discount_list = list(map(lambda tag: tag.text.replace("원", ""), discount))
print(f"할인가\n{discount_list}")
print()

# 별점
scores = aladin_soup.select(".star_score")
score_list = list(map(lambda tag: tag.text, scores))
print(f"별점\n{score_list}")
print()

aladin_info = list(zip(title_list, price_list, discount_list, score_list))
print(f"알라딘리스트\n{aladin_info}")
```
