```python
musinsa_soup = bs(chrome.page_source, "html.parser")
```

```python
# for i in musinsa_soup.select(".SnapFeedCard__Container-sc-e43ff597-0"):
#     print(i.select_one("img")["src"])

snap_images = [i.select_one("img")["src"] for i in musinsa_soup.select(".SnapFeedCard__Container-sc-e43ff597-0")]
len(snap_images)
```

```python
import os

base_path = r"E:\kdt_0900_sal\python\resource"
folder_name = r"musinsa"

musinsa_path = os.path.join(base_path, folder_name)

# 폴더 없으면 생성
os.makedirs(musinsa_path, exist_ok=True)
```

```python
for image_url in snap_images:
    file_name = image_url.split("/")[-1].split("?")[0]
    file_path = os.path.join(musinsa_path, file_name)

    response = requests.get(image_url, stream=True)

    with open(file_path, "wb") as f:
        for chunk in response.iter_content(chunk_size=8192):
            f.write(chunk)

print("파일 저장 완료")
```

```python
review_text_list = [i.text for i in review_text if i.text]

# for i in review_text:
#     print(i.text)

review_text_list
```
