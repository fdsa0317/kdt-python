## 삼항 연산자(if else)
- 조건식이 True 일 때 값 if 조건식 else 조건식이 False일 때 값

```python
num = 10

result = "짝수" if num % 2 == 0 else "홀수"

print(result)
```

```python
#  조건) 
#  나이는 임의로 설정, 삼항 연산자
#  ** 중요 ** 값을 모두 변수에 담아보기

#  실습)
#  1. 나이가 19살 이상이면 "입장 가능" 출력
#  19살 미만이면 "입장 불가" 출력
#  2. 나이가 짝수면 "짝수" 출력
#  나이가 홀수면 "홀수" 출력
#  3. 나이가 16살이면 "당첨" 출력
#  아니면 "꽝입니다" 출력

#  예상 결과)
#  콘솔 결과: [본인이름]님은 [입장 가능]
#  [나이], [홀수], [꽝입니다].  
```

```python
age = 25
name = "홍길동"

age1 = "입장 가능 " if age >= 19 else "입장 불가"
age2 = "짝수" if age % 2 == 0 else "홀수"
age3 = "당첨" if age == 16 else "꽝입니다"


print(name + "님은 " + str(age1) + str(age) + ", " + str(age2) + ", " + str(age3))
```

```python
age = 21
name = "홍길동"
is_adult = age >= 19
pass_message = "입장 가능"
not_pass_message = "입장 불가"
even_message = "짝수"
odd_message = "홀수"
lucky_message = "당첨"
unlucky_message = "꽝"

adult_result = pass_message if is_adult else not_pass_message
is_even = age % 2 == 0
even_result = even_message if is_even else odd_message
is_lucky = age == 16
lucky_result = lucky_message if is_lucky else unlucky_message

result = name + "님은 " + adult_result + "\n" + str(age) + ", " + even_result + ", " + lucky_result
print(result)
```

## 포맷 스트링(format string)
- f"{변수명}"
- 문자열 안에 템플릿 `{}`을 생성하고, 변수를 사용할 수 있게 만든 문법

```python
# 탐욕 알고리즘(greedy)
# 사용자의 거스름돈의 최소 동전 개수 구하기
# 동전의 종류 1000, 500, 100, 10
# 12790원

# 예상 출력 결과
# 동전의 개수:
# 1000원: 12개
# 500원: 1개
# 100원: 2개
# 10원: 9개
```

```python
money1 = int(12000 / 1000)
money2 = int(500 / 500)
money3 = int(200 / 100)
money4 = int(90 / 10)

print(f"동전의 개수:\n1000원: {money1}개\n500원: {money2}개\n100원: {money3}개\n10원: {money4}개")
```

```python
money = 12790
coin_1000 = 1000
coin_500 = 500
coin_100 = 100
coin_10 = 10

count_1000 = money // coin_1000
money %= coin_1000
count_500 = money // coin_500
money %= coin_500
count_100 = money // coin_100
money %= coin_100
count_10 = money // coin_10
money %= coin_10

result = f"동전의 개수:\n1000원: {count_1000}개\n500원: {count_500}개\n100원: {count_100}개\n10원: {count_10}개"
print(result)
```

```python
#  심화
#  주차 요금 구하기
#  1. 기본 주차 요금은 시간당 1,000원
#  2. 주말이면 총 요금의 20% 추가 요금이 붙는다
#  3. 이용 시간이 5시간 초과이면 총 요금에서 10% 할인
#  4. 최종 요금을 계산하여 출력한다
 
#  주차시간과 주말여부는 임의로 넣는다.
#  모든 값은 변수에 담아서 사용한다.
```

```python
hour = 6
weekend = "평일"
total = hour * 1000
total = (total + (total * 0.2)) if weekend == "주말" else total 
total = (total - (total * 0.1)) if hour > 5 else total

print(int(total))
```

```python
hour = 4
is_weekend = False
total = hour * 1000
total *= 1 + is_weekend * 0.2
total *= 1 - (hour > 5) * 0.1
total = int(total)

print(f"주차 시간: {hour}시간\n주차 비용: {total}원")
```
