```python
 # 실습
# 유저(User) 클래스를 생성하고 아래 코드를 완성하시오
# 유저는 이름(name), 돈(money)를 가지고 있다. 돈은 모두 10000원씩 존재한다
# 유저는 출금(withdraw) 메서드를 통해 유저의 돈을 출금할 수 있다.
# 아래에 코드를 완성하시오.
```

```python
class User:
    def __init__(self, name, money = 10000):
        self.name = name
        self.money = money

    def withdraw(self, money):
        if self.money < money:
            print("출금할 수 없습니다")
            return -1
        self.money -= money
        print(f"{self.name}님의 남은 돈은 {self.money}원")
```

```python
class User:
    def __init__(self, name, money = 10000):
        self.name = name
        self.money = money

    def info(self):
        print(f"{self.name}님의 잔액: {format(self.money, "_")}원")

    def withdraw(self, withdraw_money):
        # 내가 가진 돈보다 많은 경우 return
        if withdraw_money > self.money:
            print(f"잔액이 부족합니다.\n잔액: {format(self.money, "_")}원")
            return -1     # -1 하는 이유: 출력엔 문제 없지만 오류를 확인해야할 때
            
        # 출금이 가능한 경우
        # 1. 내가가진 돈을 빼야된다.
        # self.money = self.money - withdraw_money
        self.money -= withdraw_money
    
        # 2. 돈을 반환해야 한다.
        print(f"출금 완료\n잔액:{format(self.money, "_")}")
        return withdraw_money

hong = User("홍길동")
jang = User("장보고")

hong.info()
jang.info()

# 홍길동 출금
hong.withdraw(15000)
```

```python
# 실습, 아래 코드를 설계하시오.
# 1. 사물함(Locker) 클래스가 존재한다.
# 2. 사물함은 주인(owner)과 비밀번호(password), 사물함의 물건(items)가 존재한다.
# 3. 물건을 추가하는 메서드는(add_item) 사물함의 물건에 전달받은 물건을 추가한다.
# 4. 물건을 꺼내는 메서드는(get_item) 사물함의 물건에서 요청받은 물건을 꺼내고 반환한다.
# 5. 사물함은 추가하거나, 꺼낼 때 비밀번호가 일치해야한다.
```

```python
class Locker:
    def __init__(self, owner, password, items = []):
        self.owner = owner
        self.password = password
        self.items = items

    def add_item(self, password, item):
        if password == self.password:
            self.items.append(item)
            print(f"{self.owner}님의 {item}를 추가했습니다.\n사물함 물건: {', '.join(self.items)}")
        else:
            print("비밀번호를 다시 입력해주세요")

    def get_item(self, password, item):
        if item not in self.items:
            print(f"{item}가 없습니다\n현재 사물함 물건: {', '.join(self.items)}")
            return
        
        if password == self.password:
            self.items.remove(item)
            print(f"{self.owner}님의 {item}를 꺼냈습니다.\n사물함 물건: {', '.join(self.items)}")
        else:
            print("비밀번호를 다시 입력해주세요")

    def show_item(self):
        print(f"{self.owner}님의 사물함 목록: {', '.join(self.items)}")
```

```python
class School:
    def __init__ (self, name, students):
        self.name = name
        self.students = students

    def get_total_count(self):
        print(f"{self.name}학교의 학생 수는{len(self.students)}명 입니다.")

    def print_student_rank(self):
        result = []
        for name, info in self.students.items():
            total = info["kor"] + info["eng"] + info["math"]
            result.append((name, total))
        result.sort(key=lambda student: student[1], reverse=True)

        for rank, student in enumerate(result, 1):   # enumerate(자료형, 시작 번호)
            print(f"{rank}등 {student[0]}: {student[1]}점")
```

```python
# 기초 풀이
# 실습 아래코드를 설계하시오.
# 1. 상점(Store) 클래스가 존재한다.
# 2. 상점은 이름(name), 상품들(products)를 가지고 있다.
# 3. 상점은 상품 중 구매할 상품을 전달받아서 판매하는 (sell)함수를 존재한다.
# 4. 상점에는 상품을 추가할 수 있는 함수(insert_product)가 존재한다.
# 5. 상점에는 상품들을 여러개 한 번에 추가할 수 있는 함수(insert_products)가 존재한다.
```

```python
class Store:
    def __init__(self, name, products = []):
        self.name = name
        self.products = products

    def sell(self, products):
        if products not in self.products:
            return print(f"{products}는 현재 품절 입니다.\n현재 상품 목록: {', '.join(self.products)}")          
        else:
            self.products.remove(products)
        print(f"{products} 상품이 판매 되었습니다.\n현재 상품 목록: {', '.join(self.products)}")

    def insert_product(self, products):
        self.products.append(products)
        print(f"{products}상품이 추가 되었습니다.\n현재 상품 목록: {', '.join(self.products)}")

    def insert_products(self, products = []):
        for i in products:
            self.products.append(i)
        print(f"{', '.join(products)}상품들이 추가 되었습니다.\n현재 상품 목록: {', '.join(self.products)}")
```

```python
class Store:
    def __init__(self, name, products = []):
        self.name = name 
        self.products = products

    def sell(self, product):
        # 상품이 없다면, 판매하면 안된다.
        if not product in self.products:
            print(f"{product} 상품을 판매할 수 없습니다")
            self.show_products()
            return 

        # 상품이 있다면
        print(f"{product} 판매 완료")
        sell_item = self.products.pop(self.products.index(product))
        self.show_products()
        return sell_item

    def insert_product(self, product):
        if product in self.products:
            print(f"이미 상품이 존재합니다")
            return 
        self.products.append(product)
        print(f"{product} 상품 추가 완료")
        self.show_products()

    def insert_products(self, *args):
        # self.products = self.products + list(args)
        self.products += list(set(args))
        print(f"{", ".join(args)} 상품 추가 완료")
        self.show_products()

    def show_products(self):
        print("현재 상품 목록:\n" + ", ".join(self.products))

market = Store("중고나라", ["아이폰", "헤드셋", "신발", "옷"])
print(market.sell("아이폰"))
market.insert_product("모자")
market.insert_products("모자", "핸드폰", "가방", "모자", "모자", "모자")
```

```python
# Phone클래스를 상속
# 이름(name), 가격(price)
# 전화기능(call), 인터넷(browse)

# Phone2G(Phone), Phone3G(Phone2G), Phone4G(Phone3G)
```

```python
class Phone:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def call(self):
        print("전화하는 기능")

    def browse(self):
        print("인터넷 안됨")
```

```python
# Phone2G(Phone), Phone3G(Phone2G), Phone4G(Phone3G)
```

```python
class Phone2G(Phone):
    def __init__(self, name, price):
        super().__init__(name, price)

    # 오버라이딩(재정의)
    def browse(self):
        print("2G 인터넷")

    # 2G의 +@ == 자식만의 코드
    def sms(self):
        print("문자 발송")
```

```python
class Phone3G(Phone2G):
    def __init__(self, name, price, brand):
        super().__init__(name, price)
        # 자식만의 코드
        self.brand = brand

    # 오버라이딩(재정의)
    def browse(self):
        print("3G 인터넷")

    # 3G의 +@ == 자식만의 코드
    def stream_video(self):
        print("영상통화")
```

```python
# Phone4G(Phone3G)
# AI탑재기능, 인터넷 무료

class Phone4G(Phone3G):
    def __init__(self, name, price, brand):
        super().__init__(name, price, brand)

    # 오버라이딩(재정의)
    def browse(self):
        print("4G 인터넷")

    # 4G의 +@ == 자식만의 코드
    def add_ai(self):
        print("AI 기능")

    # 4G의 +@ == 자식만의 코드
    def internet_free(self):
        print("인터넷 무료")
```

```python
phone2G = Phone2G("2G", 10000)
phone2G.browse()
phone2G.call()
phone2G.sms()

phone3G = Phone3G("3G", 100000, "삼성")
phone3G.browse()
phone3G.call()
phone3G.sms()
phone3G.stream_video()

phone4G = Phone4G("4G", 1000000, "애플")
phone4G.browse()
phone4G.call()
phone4G.sms()
phone4G.stream_video()
phone4G.internet_free()
phone4G.add_ai()
```

```python
class Knight:
    def __init__(self, name, hp):
        self.name = name
        self.hp = hp

    def attack(self):
        print("검으로 공격한다.")

class ArchMage:
    def __init__(self, name, hp, mp):
        self.name = name
        self.hp = hp
        self.mp = mp

    def magic(self):
        print("마법으로 공격한다.")

class User(Knight, ArchMage):
    def __init__(self, name, hp, mp):
        Knight.__init__(self, name, hp)
        ArchMage.__init__(self, name, hp, mp)

user = User("홍길동", 100, 50)

print(user.name)
print(user.hp)
print(user.mp)
```

```python
print(isinstance(animal, Animal)) # animal은 Animal(부모)타입이 맞다
print(isinstance(animal, Person)) # animal은 Person(자식)타입이 아니다

print(isinstance(person, Animal)) # person은 Animal(부모)타입이 맞다
print(isinstance(person, Person)) # person은 Person(자식)타입이 맞다
# 모든 자식은 부모타입이다.
```

## 마커(Maker)

```python
# 동물들
# 초식동물, 육식동물, 잡식동물

# 초식동물
class HerbivoreMaker:
    pass

# 육식동물
class CarnivoreMaker:
    pass

class Animal:
    def __init__(self, name):
        self.name = name

class Bear(Animal, CarnivoreMaker):
    def __init__(self, name):
        super().__init__(name)
        
class Cow(Animal, HerbivoreMaker):
    def __init__(self, name):
        super().__init__(name)
        
class Dog(Animal):
    def __init__(self, name):
        super().__init__(name)
        
class Tiger(Animal, CarnivoreMaker):
    def __init__(self, name):
        super().__init__(name)
        
class Cat(Animal):
    def __init__(self, name):
        super().__init__(name)
```

```python
animals = [Bear("곰"), Cow("소"), Dog("강아지"), Tiger("호랑이"), Cat("고양이")]
for animal in animals:
    if isinstance(animal, CarnivoreMaker):
        print(animal.name)
```

## 객체간의 프로그래밍

```python
# 은행, 유저, 송금, 출금
class Bank:
    def __init__(self, name, bank_money = 1000000):
        self.name = name
        self.bank_money = bank_money

    # 입금
    def deposit(self, user, money):
        # 은행에도 1만원 입금
        self.bank_money += money
        
        # user돈도 1만원 출금
        user.user_money -= money
        
    # 출금
    def withraw(self, user, money):
        # 은행에 돈만큼 -
        self.bank_money -= money
        # 유저에게 돈만큼 +
        user.user_money += money

class User:
    def __init__(self, name, user_money = 0):
        self.name = name
        self.user_money = user_money

kb = Bank("국민은행")
hong = User("홍길동", 100000)
```

```python
kb.deposit(hong, 10000)
print(hong.user_money)

kb.withraw(hong, 50000)
print(hong.user_money)
```

```python
# 쿠폰 할인율
# 상점(Store), 고객(Customer)
# 판매(sell)
# 상품: 모자 - 10000

class Store:
    def __init__(self, name):
        self.name = name
        self.product = {"name": "모자", "price": 10000}

    def sell(self, customer, count):
        price = self.product["price"]
        discounted_price = price - int(self.product["price"] * (customer.discount / 100))
        customer.money -= price * count
        print(f"원래 상품의 가격: {price}원, 할인된 가격: {discounted_price}")

class Customer:
    def __init__(self, name, money, discount):
        self.name = name
        self.money = money
        self.discount = discount
```

```python
store = Store("인터넷마켓")
hong = Customer("홍길동", 50000, 15)
jang = Customer("장보고", 100000, 80)

store.sell(hong, 1)
store.sell(jang, 1)
```

## 실습

```python
# 마켓 문제
# - 마켓(Market) 클래스
# 1) 마켓은 이름(name), 판매 상품(product), 수입(income)의 필드를 갖는다.
# 2) 판매 상품은 다음과 같다
# - 이름(name)은 “토마토”이며, 토마토의 가격(price)은 3000원, 재고(stock)은 10개이다.
# 3) 마켓은 유저에게 상품을 판매할 수 있는 함수(sell)을 가지고 있다.
# - 유저가 가지고 있는 돈이 상품의 판매 가격보다 적다면 판매할 수 없다.
# - 유저가 구매하는 개수보다 재고의 개수가 적다면 판매할 수 없다.
# - 상품 판매시 유저가 가지고 있는 할인율이 적용되어 판매가 된다.

# ex) 유저의 할인율이 10%라면 2700원이 결제된다.
# - 판매된 상품의 모든 수익금은 income에 누적된다.


# - 구매자(Buyer) 클래스
# 1) 구매자는 이름(name), 돈(money), 할인율(discount)의 필드를 갖는다.
# 2) 총 3명의 Buyer를 각각 객체화
# - 홍길동, 15000, 5
# - 장보고, 10000, 20
# - 이순신, 30000, 50
 
# 구매자의 할인율을 적용받아 마켓에서 토마토를 구매를 아래와 같이 구매하였다.
# 마켓의 총 수익과 각 구매자의 잔액을 출력하시오.
# - 홍길동 6개 구매
# - 장보고 2개 구매
# - 이순신 8개 구매
```

```python
class Market:
    def __init__(self, name, product, income = 0):
        self.name = name
        self.product = {"name": "토마토", "price": 3000, "stock": 10}
        self.income = income

    def sell(self, buyer, count):   # 위에서 아래로 읽히니까, 할인가, 할인율 등을 먼저 구한 후 조건으로 검사하고, 마지막에 결과(누적값, 출력 등)
        # 토마토의 가격
        price = self.product["price"]

        # 할인율 / 산 갯수만큼의 할인받은 가격
        discounted_price = price - int(self.product["price"] * (buyer.discount / 100))   # discounted_price : 할인가
        total_discounted = discounted_price * count   # total_discounted : 할인율 포함된 구매 가격
        total_price = price * count  # 할인 받지 않은 원가

        # 구매하려는 가격보다 가지고 있는 돈이 부족할 때 / 현재 재고 수보다 사려는 개수가 많을 때
        if buyer.money < total_discounted:
            print(f"돈이 부족합니다.\n{buyer.name}님의 잔액:{buyer.money}")
            return
        if count > self.product["stock"]:
            print(f"상품이 부족합니다.\n 현재 재고: {self.product["stock"]}")
            return

        # 현재 고객 돈에서 구매가격 빼기 / 인컴에 수익 더하기 / 남아있는 재고에서 구매한 개수 빼기
        buyer.money -= total_discounted
        self.income += total_discounted
        self.product["stock"] -= count
        print(f"원가: {total_price}원, 할인받은 가격: {discounted_price}원, 총 구매 가격: {total_discounted}원으로 판매 완료\n{buyer.name}님의 남은 잔액: {buyer.money}")

class Buyer:
     def __init__(self, name, money, discount):
         self.name = name
         self.money = money
         self.discount = discount
```

```python
market = Market("마트", "토마토")

hong = Buyer("홍길동", 15000, 5)
jang = Buyer("장보고", 10000, 20)
lee = Buyer("이순신", 30000, 50)
```

```python
# - 홍길동 6개 구매
# - 장보고 2개 구매
# - 이순신 8개 구매
```

```python
market.sell(hong, 6)
print()
market.sell(jang, 2)
print()
market.sell(lee, 8)
print()
print(f"마트의 총 수입: {market.income}")
```

```python
print(f"홍길동님의 잔액: {hong.money}원\n장보고님의 잔액: {jang.money}원\n이순신님의 잔액: {lee.money}원\n마트의 총 수익: {market.income}원")
```
