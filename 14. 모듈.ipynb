```python
import math

# 3.14
# .ceil(): 올림
print(math.ceil(3.14))
# .floor(): 내림
print(math.floor(3.99))
# round(): 반올림
print(round(2.5))
# .factorial() == 5! : 5x4x3x2x1
print(math.factorial(5))
# .pi: 원주율
print(math.pi)
```

```python
# .choice(): 시퀀스 자료구조 중에 1개를 선택하여 반환

# 추첨
students = ["a", "b", "c", "d", "e", "f", "g"]

print(random.choice(students))
```

```python
students = ["a", "b", "c", "d", "e", "f", "g"]

print(random.sample(students, 4))
```

```python
from collections import Counter

# 리스트 또는 시퀀스 객체에서 개수를 세는 객체

fruits = "apple banana apple banana apple melon".split()
fruit_count = Counter(fruits)

print(fruit_count)
print(fruit_count["apple"])
```

```python
text = "hello, world, python"
text_count =Counter(text)

print(text_count["o"])
```

```python
# key가 없어도 오류가 나지 않음
print(fruit_count["a"])
# key가 없으면 오류
print({'apple': 3, 'banana': 2, 'melon': 1}["a"])
```

```python
import time

# 시작 시간
start_time = datetime.now()
# .sleep(5): 5초동안 코드가 멈춤
time.sleep(5)

# 종료 시간
end_time = datetime.now()

# 시작 - 종료
elapsed = end_time - start_time

print(f"경과 시간: {elapsed}")
```
