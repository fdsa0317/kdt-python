```python
# 마켓(Market) 클래스
# 1) 마켓은 이름(name), 판매 상품(product), 수입(income)의 필드를 갖는다.
# 2) 판매 상품은 다음과 같다
# - 이름(name)은 “토마토”이며, 토마토의 가격(price)은 3000원, 재고(stock)은 10개이다.
# 3) 마켓은 유저에게 상품을 판매할 수 있는 함수(sell)을 가지고 있다.
# - 유저가 가지고 있는 돈이 상품의 판매 가격보다 적다면 판매할 수 없다.
# - 유저가 구매하는 개수보다 재고의 개수가 적다면 판매할 수 없다.
# - 상품 판매시 유저가 가지고 있는 할인율이 적용되어 판매가 된다.

# ex) 유저의 할인율이 10%라면 2700원이 결제된다.
# - 판매된 상품의 모든 수익금은 income에 누적된다.
tomato = {"name": "토마토", "price": 3000, "stock": 10}

class Market:
    def __init__(self, name, product, income = 0):
        self.name = name
        self.product = product
        self.income = income

    def sell(self, buyer, count):
        price = self.product["price"]

        discount = price * buyer.discount / 100
        discount_price = price - discount
        total_price = discount_price * count

        if buyer.money < total_price:
            print(f"현재 잔액이 부족합니다.\n상품 가격: {round(total_price)}\n{buyer.name}님의 잔액: {buyer.money}")
            return

        if self.product["stock"] < count:
            print(f"현재 상품 재고가 부족합니다.\n현재 남은 재고:{self.product["stock"]}")
            return

        buyer.money -= total_price
        self.income += total_price
        self.product["stock"] -= count

        print(f"{buyer.name}님에게 판매 완료!\n상품원가: {price}\n할인가: {round(discount_price)}\n구매가: {round(total_price)}\n감사합니다.")

market = Market("롯데마트", tomato)
```

```python
# 구매자(Buyer) 클래스
# 1) 구매자는 이름(name), 돈(money), 할인율(discount)의 필드를 갖는다.

# 2) 총 3명의 Buyer를 각각 객체화
# - 홍길동, 15000, 5
# - 장보고, 10000, 20
# - 이순신, 30000, 50
 
# 구매자의 할인율을 적용받아 마켓에서 토마토를 구매를 아래와 같이 구매하였다.
# 마켓의 총 수익과 각 구매자의 잔액을 출력하시오.
# - 홍길동 6개 구매
# - 장보고 2개 구매
# - 이순신 8개 구매
class Buyer:
    def __init__(self, name, money, discount):
        self.name = name
        self.money = money
        self.discount = discount

hgd = Buyer("홍길동", 15000, 5)
jbg = Buyer("장보고", 10000, 20)
lss = Buyer("이순신", 30000, 50)
```

```python
market.sell(hgd, 6)
market.sell(jbg, 2)
market.sell(lss, 8)

print(f"홍길동 남은 잔액: {round(hgd.money)}\n장보고 남은 잔액: {round(jbg.money)}\n이순신 남은 잔액: {round(lss.money)}")
```

```python
print(f"마트 총 수익: {round(market.income)}")
```
