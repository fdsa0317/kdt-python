## 커스텀 예외 생성
- 모든 예외(오류)의 부모(Exception) 클래스를 상속받는 것을 의미한다

```python
class TextException(Exception):
    pass

try:
    text = input("채팅을 입력하세요")
    
    word_list = ["바보", "멍청이", "무친사람", "등신"]
    for word in word_list:
        if word in text:
            raise TextException("맘아프게 말하지마..😢")

except TextException:
    print("비속어를 사용하지 마세요😡")

except Exception:
    print("알 수 없는 오류 발생")
```

```python
class TextException(Exception):
    def __init__(self, message, state, code):
        super().__init__(message)
        self.state = False
        self.code = 400
```

```python
# 국어, 영어, 수학 점수를 받아서 총점과 평균을 구해주는 프로그램
# 0 ~ 100점 까지만 입력할 수 있습니다.
# 이름을 입력할 수도 있습니다.
# 학번을 입력할 수도 있습니다.
# 평균의 소숫점 자리를 입력받아 처리할 수도 있습니다.
# 예외가 발생한다면 try ~ except 문법으로 해결하세요.
```

```python
class Number(Exception):
    pass

try:
    kor = int(input("국어 점수:"))
    if  0 > kor or kor > 100:
        raise Number

    eng = int(input("영어 점수:"))
    if  0 > eng or eng > 100:
        raise Number

    math = int(input("수학 점수:"))
    if  0 > math or math > 100:
        raise Number

except Number:
    print("점수는 0 ~ 100까지만 입력 가능합니다")

except ValueError:
    print("점수(숫자)만 입력해주세요")

else:
    total = kor + eng + math
    average = total / 3
    
    print()
    print(f"국어: {kor}\n영어: {eng}\n수학: {math}\n총 점: {total}\n평균: {round(average, 2)}")
```

```python
datas = {
    "kor": "국어 점수", 
    "eng": "영어 점수", 
    "math": "수학 점수", 
    "name": "이름", 
    "number": "학번",
    "round": "반올림 수"
}

class AlphaException(Exception):
    pass

try:
    for idx, key in enumerate(datas):
        message = f"{datas[key]}를 입력하세요.\n"
        input_value = input(message)
        
        if idx <= 2 or idx == 5:
            if input_value.isalpha():
                raise AlphaException(f"{datas[key]}는 문자를 입력할 수 없습니다.")
            else: 
                input_value = int(input_value)

        datas[key] = input_value
        
    datas["total"] = datas["eng"] + datas["kor"] + datas["math"]
    datas["average"] = round(datas["total"] / 3, datas.get("round", 0))

    for i in datas.items():
        print(i)
    
except AlphaException as e: 
    print(e)

except TypeError as e:
    print("타입이 맞지 않습니다.")

except Exception as e:
    print("알 수 없는 오류입니다.")
```
