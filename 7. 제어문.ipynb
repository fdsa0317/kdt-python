```python
num = 12

# if~if문
# 3의 배수인지 확인
if num % 3 == 0:
    print("3의 배수")
    
if num % 3 ==0:
    print("3의 배수")
```

```python
# if~else문
if num % 3 == 0:
    print("3의 배수")
    
elif num % 3 ==0:
    print("3의 배수")
```

```python
if num % 3 == 0:
    print("3의 배수입니다.")

# else: 위에 있는 모든 조건식을 만족하지 않으면
else:
    print("3의 배수가 아닙니다.")
```

```python
# 숫자 2개를 변수에 담고, 
# 두 수를 비교해서 큰 수와 작은 수를 출력하기
# 단, 두 수가 같을 수도 있음
# 만약 두 수가 같다면 "두 수는 같습니다"를 출력하기
num1 = 5
num2 = 5

if num1 > num2:
    print(f"{num1}이 {num2}보다 큽니다")
elif num1 < num2:
    print(f"{num2}이 {num1}보다 큽니다")
else:
    print("두 수는 같습니다")
```

```python
# 점수 등급
# A: 90, B: 80, C: 70, D: 60, 60미만은 과락

# 사용자에게 이름과 수학, 영어, 국어 평균 점수의 등급을 매기는 로직 구현

# 출력 예시)
# 홍길동님의 총 점은 000점, 평균은 00.00점으로 00등급입니다.

name = "홍길동"
math = 76
eng = 78
kor = 95
total = math + eng + kor
result = total / 3

if result >= 90:
    print(f"{name}님의 총 점은 {total}점, 평균은 {result:.2f}점으로 A등급입니다.")
elif result >= 80:
    print(f"{name}님의 총 점은 {total}점, 평균은 {result:.2f}점으로 B등급입니다.")
elif result >= 70:
    print(f"{name}님의 총 점은 {total}점, 평균은 {result:.2f}점으로 C등급입니다.")
elif result >= 60:
    print(f"{name}님의 총 점은 {total}점, 평균은 {result:.2f}점으로 D등급입니다.")
else:
    print(f"{name}님의 총 점은 {total}점, 평균은 {result:.2f}점으로 과락입니다.")
```

```python
name = "홍길동"
math = 71
eng = 84
kor = 99
total = math + eng + kor
average = total / 3

if average >= 90:
    grade = "A"
elif average >= 80:
    grade = "B"
elif average >= 70:
    grade = "C"
elif average >= 60:
    grade = "D"
else:
    grade = "과락"

result = f"홍길동님의 총 점은 {total}점, 평균은 {average}점으로 {grade}등급입니다."
print(result)
```
