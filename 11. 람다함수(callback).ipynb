# 콜백함수

```python
# 두 수를 더해서 출력 함수
def sum1(num1, num2, callback):
    callback(num1 + num2)

sum1(10, 20, print)
```

```python
# 성과 이름을 전달하면 풀네임을 만들고 님을 붙여서 반환
def get_full_name(last_name, first_name, callback):
    return callback(last_name + first_name)

def add_nim(name):
    return name + "님"

get_full_name("홍", "길동", add_nim)
```

```python
# 이름을 전달하면 이름을 5번 출력해주는 함수 (콜백함수 사용)
def get_name(name, callback):
    return callback(name)

def name_five(name):
    return name * 5

get_name("홍길동", name_five)
```

```python
def print_name(name):
    print(name)

def repeat(name, callback):
    for i in range(5):
        callback(name)

repeat("홍길동", print_name)
```

```python
def repeat(name, callback):
    for i in range(5):
        callback(name)

repeat("홍길동", lambda name: print(name))
```

```python
# 두 수를 더하고 2배를 곱한 결과에 10배를 만들고 반환
def add(num1, num2, callback):
    return callback(num1 + num2)

def get_square(num, callback):
    return callback(num ** 2)

def get_multiple_by_10(num):
    return num * 10

add(10, 20, lambda add_result: get_square(add_result, get_multiple_by_10))
```

# 람다 실습

```python
# 두 정수의 결과를 곱하고 출력
# 횟수를 전달하면 횟수까지의 합을 반환
# 1~100까지 짝수만 더한 뒤 결과 값의 2를 곱한 후 출력
# 두 수를 더하고 홀수인지 짝수인지 비교한 후 출력
```

```python
# 두 정수의 결과를 곱하고 출력
def get_num(num1, num2, callback):
    return callback(num1 * num2)

get_num(2, 5, lambda num_result : (num_result))
```

```python
# 두 정수의 결과를 곱하고 출력
def multiple(num1, num2, callback):
    callback(num1 * num2)

multiple(10, 20, lambda num : print(num))
```

```python
# 횟수를 전달하면 횟수까지의 합을 반환
def add_sum(num, callback):
    return callback(num)

add_sum(5, lambda num : sum(range(1, num + 1)))
```

```python
# 횟수를 전달하면 횟수까지의 합을 반환
def get_total(count, callback):
    return callback(count)

def repeat(count):
    total = 0
    for i in range(count):
        total += i + 1
    return total

get_total(5, repeat)
```

```python
# 1~100까지 짝수만 더한 뒤 결과 값의 2를 곱한 후 출력
def num(callback):
    total = 0
    for i in range(1, 101):
        if i % 2 == 0:
            total += i

    return callback(total)

num(lambda num : num * 2)
```

```python
# 1~100까지 짝수만 더한 뒤 결과 값의 2를 곱한 후 출력
def get_total_even(callback):
    callback(sum([i for i in range(0, 101, 2)]))

def multiple(num, callback):
    callback(num * 2)

get_total_even(lambda total : multiple(total, print))
```

```python
# 두 수를 더하고 홀수인지 짝수인지 비교한 후 출력
def add_sum(num1, num2, callback):
    return callback(num1 + num2)

add_sum(6, 6, lambda num : "짝수" if num % 2 == 0 else "홀수")
```

```python
# 두 수를 더하고 홀수인지 짝수인지 비교한 후 출력
def add(num1, num2, callback):
    callback(num1 + num2)

def confirm_odd_and_even(num, callback):
    callback("짝수" if num % 2 == 0 else "홀수")

add(10, 20, lambda x : confirm_odd_and_even(x, print))
```

## filter, map, reduce

### 1. map 함수
- iterable한 객체들을 순차적으로 순회하여 결과를 변환하고 반환(리턴)하는 함수

```python
li1 = [(i + 1) * 10 for i in range(10)]
li1
```

```python
for idx, value in enumerate(li1):
    li1[idx] = value * 10

print(li1)
```

```python
li2 = [(i + 1) * 10 for i in range(10)]
li2
```

```python
# map(li2)
list(map(lambda num: num * 10, li2))
```

```python
# li3 -> 홀수면 true, 짝수면 false
li3 = [i + 1 for i in range(50)]
print(li3)
```

```python
li4 = list(map(lambda num: num % 2 == 1, li3))

print(li3)
print(li4)
```

### 2. filter 함수
- iterable한 객체들을 순차적으로 순회하여 조건식을 만족하는 값들만 모아서 반환해주는 함수

```python
# filter
# 짝수만 걸러내기 

print(list(filter(lambda num: num % 2 == 0, li3)))
```

```python
li5 = [chr(ord('a') + i) for i in range(26)]
print(li5)
```

```python
# 짝수 인덱스의 값들만 걸러내고 싶을 때
# acegij,......
li6 = list(filter(lambda tu: tu[0] % 2 == 0, enumerate(li5)))
print(li6)
```

```python
li7 = list(map(lambda c: c[1], li6))
print(li7)
```

```python
# 1~10까지 리스트에 담고, 3배를 곱한 리스트로 변경
# 100~1까지 리스트에 담고, 5의 베수만 골라낸 리스트로 변경
```

```python
# 1~10까지 리스트에 담고, 3배를 곱한 리스트로 변경
li8 = [i + 1 for i in range(10)]
li9 = list(map(lambda num: num * 3, li8))
print(li9)
```

```python
# 100~1까지 리스트에 담고, 5의 베수만 골라낸 리스트로 변경
li10 = [i for i in range(100, 0, -1)]
li11 = list(filter(lambda num: num % 5 == 0, li10))
print(li11)
```

### 3. reduce 함수
- iterable한 객체들을 순차적으로 순회하여 모든 값들을 누적시켜서 반환해주는 함수

```python
from functools import reduce

# 1~10까지 모든 합을 구하기
li11 = [i + 1 for i in range(10)]
li11

reduce(lambda acc, crr: acc + crr, li11)
```

```python
# 1~10까지 리스트에 담고 값이 5보다 크다면 5배로 변경
# 1~30까지 배열에 담고 3을 곱한 뒤 출력
# 소문자 알파벳들을 담고 아래와 같이 변경 후 출력
# aBcDeF...
```

```python
# 1~10까지 리스트에 담고 값이 5보다 크다면 5배로 변경
li1 = [i + 1 for i in range(10)]
list(map(lambda num: num * 5 if num > 5 else num, li1))
```

```python
# 1~30까지 배열에 담고 3을 곱한 뒤 출력
li2 = [i + 1 for i in range(30)]
print(list(map(lambda num: num * 3, li2)))
```

```python
# 소문자 알파벳들을 담고 아래와 같이 변경 후 출력
# aBcDeF...
li3 = [chr(ord('a') + i) for i in range(26)]
li4 = [chr(ord('A') + i) for i in range(26)]

li5 = list(filter(lambda num: num[0] % 2 == 0, enumerate(li3)))
li6 = list(filter(lambda num: num[0] % 2 == 1, enumerate(li4)))

def small_big():
    for i in range(len(li5)):
        print(f"{li5[i][1]}{li6[i][1]}", end="")

small_big()
```

```python
li5 = [chr(ord('a') + i) for i in range(26)]
print(li5)

li6 = list(map(lambda c: chr(ord(c) - 32) if ord(c) % 2 == 0 else c, li5))
print(li6)
print("".join(li6))
```
