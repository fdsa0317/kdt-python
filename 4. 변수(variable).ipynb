# 변수 선언 시 주의사항
1. 되도록 선언부에 한꺼번에 선언한다 (영역 상단)
2. 초기화를 해준다

## 변수명 선언 시 주의사항
1. 문자로 시작해야 한다
2. 특수문자는 사용할 수 없다 (단, _는 허용한다)
3. 소문자로 시작한다
4. 공백을 사용할 수 없다
5. 명사로 사용한다
6. 되도록 한글은 사용하지 않는다
7. 뜻이 있는 단어를 사용한다

ex)
- a, b, c, d, e, ... (X)
- data, number, age, name, ...(O)

### 실습 문제

```python
# 변수에 담아서 출력하기
# 'cherry', 'apple', 'melon', 'banana', 'mango'
```

```python
fruits = 'cherry'
print(fruits)
fruits = 'apple'
print(fruits)
fruits = 'melon'
print(fruits)
fruits = 'banana'
print(fruits)
fruits = 'mango'
print(fruits)
```

```python
# 다음 값들을 알맞는 이름의 변수에 담고
# 정수형으로 출력하기
# 형변환 + 변수
# "a", False, "5", "10.7"
```

```python
num1 = "a"
num2 = False
num3 = "5"
num4 = "10.7"
print(int(bool(num1)),int(num2),int(num3),int(float(num4)))
```

```python
# 모든 값을 변수에 담기
# 딱 1번씨간 사용해서 100만들기
True, 0., ""
```

```python
data1 = True
data2 = 0.
data3 = ""
print(str(int(data1)) + str(int(data2)) + str(int(bool(data3))))
```
