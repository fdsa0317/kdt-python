```python
class User:
    name: str
    age: int
    phone: str

    def __init__(self, name, age, phone):
        self.name = name
        self.age = age
        self. phone = phone

    # __main__
    def __str__(self):
        return f"User(name: {self.name}, age: {self.age}, phone: {self.phone})"
```

```python
print(User("홍길동", 20, "010-1234-1234"))
```

```python
class Student(BaseModel):
    id: int
    name: str

    def __eq__(self, student):
        self.id == student.id
```

```python
student1 = Student(id=1, name="홍길동")
student2 = Student(id=1, name="홍길동")
student3 = Student(id=2, name="장보고")
student4 = Student(id=3, name="홍길동")

print(student1 == student2)
print(student1 == student3)
print(student3 == student4)
```

```python
class Person:
    def __init__(self, name):
        self.name = name
        # __repr__: 해시 주소
        print("__repr__ 호출: ", self.__repr__)

    def __eq__(self, other):
        # 같은 주소인지
        if self.__repr__() == other.__repr__():
            return True

        # 같은 타입이라면
        if isinstance(other, Person):
            # 특정 값이 같은지 비교
            return self.name == other.name

        # 한 번이라고 if에 안들어갔다면 False
        return False

class Animal:
    def __init__(self, name):
        self.name = name
```

```python
person1 = Person("홍길동")
animal = Animal("홍길동")

print(person1.__eq__(animal))
```
