```python
# 1) 이름을 5번 출력해주는 함수
def print_name(name):
    for i in range(5):
        print(name)

print_name("홍길동")
```

```python
# 2) 어떤 값을 전달하면 그 값까지 모두 더한 값을 출력해주는 함수
def add_sum(num):
    total = 0
    for i in range(num):
        total += i + 1
    return total

add_sum(15)
```

```python
# 3) 어떤 값을 전달하면 홀수인지 짝수인지 확인해서 결과를 반환해주는 함수
def get_odd_or_even(num):
    return f"{num}은 짝수입니다" if num % 2 == 0 else f"{num}은 홀수입니다"

get_odd_or_even(5)
```

```python
# 4) 어떤 값을 전달하면 실수인지 정수인지 확인해주는 함수
def get_int_or_float(num):
    return f"{num}은 정수입니다" if num % 1 == 0 and (not "." in str(num)) else f"{num}은 실수입니다"

get_int_or_float(3.0)
```

```python
# 5) 문자열을 전달하면 찾을 글자의 개수를 반환해주는 함수
# ex) my_func("abbacad", "a")
# 3 

def get_str_len(tu, idx):
    return tu.count(idx)

get_str_len("abchddd", "a")
```

```python
def get_count(text, c):
    count = 0
    for i in text:
        if i == c:
            count += 1

    return count

get_count("apple", "p")
```

```python
# 6) 문자열을 전달하면 글자를 뒤집어서 반환해주는 함수
# ex) "abcdef" -> "fedcba"

def get_str_sort(str1):
    li1 = sorted(str1, reverse=True)
    return li1
get_str_sort("abcde")
```

```python
def reverse(text):
    return text[::-1]

reverse("abcde")
```

```python
# 7) 0, 1을 마구잡이로 5개 전달했을 때
# 0의 개수와 1의 개수를 반환해주는 함수
# ex) my_func(0, 0, 1, 1, 0)
# 0의 개수 3개, 1의 개수 2개

def get_count_zero_and_one(*args):
    ones = sum(args)
    zeros = 5 - ones
    return f"0의 개수: {zeros}, 1의 개수: {ones}"

get_count_zero_and_one(1, 0, 0, 1, 0)
```

```python
# 심화
# 국어, 영어, 수학 점수 평균 구하기
# 자릿수는 2번째 자리부터 반올림(기본), 단 자릿수를 입력하면 자릿수대로 반올림한다.
# 사용자의 정보가 입력되면 같이 출력하기
# round()
```

```python
def std_average(kor, eng, math, **keywards):
    if "name" in keywards:
        return f"{keywards["name"]}님의 평균은 {round((kor + eng + math) / 3, 2)}점 입니다"

    return f"평균은 {round((kor + eng + math) / 3, 2)}점 입니다"

std_average(99, 88.5, 97, name="홍길동")
```

```python
def std_average(**keywards):
    # ex) ("kor", 0) -> "kor" 입력 안했을 때 0 반환, 0없으면 None을 반환해서 타입에러
    kor, math, eng = keywards.get("kor", 0), keywards.get("math", 0), keywards.get("eng", 0) 
    average = (kor + math + eng) / 3
    average = round(average, 2)
    if "name" in keywards:
        return f"{keywards["name"]}님의 평균은 {average}점 입니다"

    return f"평균은 {average}점 입니다"

std_average(kor=95, math=88.5, eng=97, name="홍길동")
```

```python
def get_average(**keywards):
    kor, eng, math = keywards.get("kor", 0), keywards.get("eng", 0), keywards.get("math", 0)
    total = kor + eng + math
    average = total / 3
    
    average = round(average, keywards.get("round", 2))
    if "name" in keywards:
        name = keywards["name"]
        return f"{name}님의 평균은: {average}점"

    return f"평균은: {average}점"

get_average(kor = 66, eng = 54, math = 98)
```

```python
# 두 값의 사칙연산 중 계산하고 싶은 값만 리턴해주기
# calc가 있을 때에만 계산해서 리턴, 없을 때에는 (10, 20)을 리턴

# 1. 두 개의 변수는 따로 변수에 받기
# 2. 부호는 따로 keywards 문법으로 받기

# ex) 10, 20, "+"

def calc(num1, num2, **keywards):
    if keywards.get("calc") == "+":
        return num1 + num2
    elif keywards.get("calc") == "-":
        return num1 - num2
    elif keywards.get("calc") == "*":
        return num1 * num2
    elif keywards.get("calc") == "/":
        return round((num1 / num2), 2)
    else:
        return (num1, num2)

calc(1, 3)
```

```python
def calculator(num1, num2, **keywards):
    if not "calc" in keywards:
        return num1, num2

    calc = keywards["calc"]
    calc_set = {"+", "-", "*", "/", "%"}
    is_confirm = calc in calc_set

    if not is_confirm:
        return "부호는 +, -, *, /, %만 사용할 수 있습니다."

    return eval(f"{num1}{calc}{num2}")
        
        
calculator(10, 20, calc="+")
```

```python
# 몇 개를 전달하더라도 가장 큰 값과 가장 작은 값을 반환해주는 함수
# my_func(10, 100, 15, 20, 25, 10)

# 가장 큰 값:100, 가장 작은 값: 10

# 심화) 10개이상으로 전달하면 "10개 이하만 가능합니다."
# 심화) count=7를 전달하면 8개까지만 받을 수 있도록 처리하기
```

```python
def num_max_and_min(*args):
    num1 = max(args)
    num2 = min(args)
    result = f"가장 큰 값: {num1}, 가장 작은 값: {num2}"

    # 심화) 10개이상으로 전달하면 "10개 이하만 가능합니다."
    if len(args) > 10:
        result = "10개 이하만 가능합니다."

    return result

num_max_and_min(10, 50, 60, 8, 69, 10)
```

```python
# 심화) count=7를 전달하면 8개까지만 받을 수 있도록 처리하기
def num_max_and_min(*args, **keywards):
    num1 = max(args)
    num2 = min(args)
    result = f"가장 큰 값: {num1}, 가장 작은 값: {num2}"

    if keywards.get("count")  == 7 and len(args) > 8:
        result = "8개까지만 가능합니다"


    return result

num_max_and_min(10, 50, 60, 8, 69, 10, 50, 60, count=7)
```

```python
def get_min_and_max(*args):
    min_num = args[0]
    max_num = args[0]

    for i in args[1:]:
        if i < min_num:
            min_num = i
        
        if i > max_num:
            max_num = i
    return f"가장 큰 값: {max_num}, 가장 작은 값: {min_num}"

get_min_and_max(10, 100, 15, 20, 27, 5)
```
