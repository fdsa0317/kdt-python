```python
# 1) 1~10까지 누적합 구하기
total = 0

for i in range(10):
    num = i + 1
    total += num
print(total)
```

```python
# 2) 1~50까지 짝수만 리스트에 담고 리스트 안에 값에 3을 곱한 값으로 변경 후 출력하기
li1 = [(i + 1) * 2 for i in range(25)]
print(li1)

for i in range(len(li1)):
    li1[i] *= 3

print(li1)
```

```python
# 3) 1~10까지 리스트에 담고 값이 5보다 크다면 값을 2배로 변경 후 원본 리스트을 출력하기
li1 = [i + 1 for i in range(10)]
print(li1)

for i in enumerate(li1):
    idx, value = i

    if value > 5:
        li1[idx] = value * 2

print(li1)
```

```python
li1 = [(i + 1) * 2 if i + 1 > 5 else i + 1 for i in range(10)]
print(li1)
```

```python
# 4) 1~30까지 리스트에 담고 홀수 값들만 "*"를 붙여서 출력하기
# ex) 1*3*5*7*9...

li1 = [i for i in range(1, 30, 2)]
print(li1)

for i in li1:
    print(i, end="*")
```

```python
# 심화
# 연속되는 글자의 개수 중 가장 많이 반복되는 글자 찾기
# 단, 같은 횟수라면 제일 먼저 들어오는 값을 리턴한다.
# 예상 출력결과
# text = "abcccddeaddddadwefda"
# 제일 많이 반복되는 글자 : d
# 횟수 : 4

text = "abcccddeaddddadwefda"

most_char = text[0]
max_count = 1

current_char = text[0]
current_count = 1

for i in range(1, len(text)):
    print(text[i])

    if text[i] == current_char:
        current_count += 1
    else:
        current_char = text[i]
        current_count = 1

    if current_count > max_count:
        most_char = current_char
        max_count = current_count

result = f"가장 연속된 글자가 많은 문자형: {most_char}\n횟수: {max_count}"
print(result)
```

```python
# 1) 100~1까지 출력
li = []

for i in range(100):
    num = i + 1
    li.append(num)
li.sort(reverse=True)
print(li)
```

```python
# 4) A~F까지 출력

for i in range(65, 71):
    print(chr(i), end="")
```

```python
# 5) A~F까지 C를 제외하고 출력

for i in range(65, 71):
    if i == 67:
        continue
    print(chr(i), end="")
```

```python
# 6) aBcDeF...Z 출력

li = []
li1 = []

for i in range(97, 123):
    if i % 2 == 1:
        a = chr(i)
        li.append(a)
for j in range(65, 91):
    if j % 2 == 0:
        b = chr(j)
        li1.append(b)
for i in range(len(li)):
    print(f"{li[i]}{li1[i]}", end="")
```
