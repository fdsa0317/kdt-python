```python
# 두 값의 사칙연산 중 계산하고 싶은 값만 리턴해주기
# calc가 있을 때에만 계산해서 리턴, 없을 때에는 (10, 20)을 리턴

# 1. 두 개의 변수는 따로 변수에 받기
# 2. 부호는 따로 keywards 문법으로 받기

# ex) 10, 20, "+"

def calc(num1, num2, **keywards):
    calc = keywards.get("calc")
    
    if calc == "+":
        return num1 + num2

    elif calc == "-":
        return num1 - num2

    elif calc == "*":
        return num1 * num2

    elif calc == "/":
        return num1 / num2

    else:
        return (num1, num2)
```

```python
# 실습, 아래 코드를 설계하시오.
# 1. 사물함(Locker) 클래스가 존재한다.
# 2. 사물함은 주인(owner)과 비밀번호(password), 사물함의 물건(items)가 존재한다.
# 3. 물건을 추가하는 메서드는(add_item) 사물함의 물건에 전달받은 물건을 추가한다.
# 4. 물건을 꺼내는 메서드는(get_item) 사물함의 물건에서 요청받은 물건을 꺼내고 반환한다.
# 5. 사물함은 추가하거나, 꺼낼 때 비밀번호가 일치해야한다.
```

```python
class Locker:
    def __init__(self, owner, password, items = []):
        self.owner = owner
        self.password = password
        self.items = items


    def add_item(self, password, item):
        if password == self.password:
            self.items.append(item)

        print(f"사물함에 {item}을 추가했습니다")
        

    def get_item(self, password, item):
        if password == self.password:
            self.items.remove(item)

        print(f"사물함에서 {item}을 꺼냈습니다")
```

```python
class Market:
    def __init__(self, name, product, income = 0):
        self.name = name
        self.product = {"name": "토마토", "price": 3000, "stock": 10}
        self.income = income

    def sell(self, buyer, count):
        price = self.product["price"]

        discount = price * buyer.discount / 100
        discount_price = price - discount
        total_price = discount_price * count

        if buyer.money < total_price:
            print(f"상품을 구매할 수 없습니다/n현재 잔액: {buyer.money}")
            return

        if self.product["stock"] < count:
            print(f"수량이 부족합니다.\n현재 남은 수량:{self.product["stock"]}")
            return

        buyer.money -= total_price
        self.income += total_price
        self.product["stock"] -= count

        print(f"총 구매 가격: {int(total_price)}원으로 판매 완료\n{buyer.name}님의 남은 잔액: {int(buyer.money)}")
```

```python
class Animal:
    def __init__(self, name, eat):
        self.name = name
        self.eat = eat

class Cat(Animal):
    def __init__(self, name, eat):
        super().__init__(name, eat)

    def action(self, action):
        print(f"{self.name}은 밥을 {self.eat}하고, {action}를 했다.")
```

```python
# 사용자가 숫자를 5개 입력했을 때,
# 그중에서 짝수의 개수와 홀수의 개수를 출력하는 프로그램을 작성하세요.
count1 = 0
count2 = 0

for i in range(5):
    num = int(input("숫자를 입력하세요: "))

    if num % 2 == 0:
        count1 += 1
    else:
        count2 += 1

print(f"짝수: {count1}\n홀수: {count2}")
```

```python
# 1부터 사용자가 입력한 숫자까지의 합을 구하세요.

num = int(input("숫자를 입력하세요: "))
total = 0

for i in range(1, num + 1):
    total += i
print(total)
```
